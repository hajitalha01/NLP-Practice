```python
# Import the function used to download and load datasets
from datasets import load_dataset

# Import Pandas for working with tabular data
import pandas as pd

# Import the regular expression module for text cleaning
import re

# Import the string module for handling punctuation
import string

# Import NLTK for natural language processing
import nltk
```

**Load the IMDB dataset**

```python
# Download and load the IMDB movie review dataset
dataset = load_dataset("stanfordnlp/imdb")

# Convert the training split into a Pandas DataFrame
df = dataset["train"].to_pandas()

# Display the first five rows of the dataset
print(df.head())

# Display the number of rows and columns
print(df.shape)
```

**Select a smaller dataset for practice**

```python
# Select 5000 random reviews for faster processing
df = df.sample(n=5000, random_state=42)

# Reset the row indexes after sampling
df = df.reset_index(drop=True)

# Display the dataset dimensions
print(df.shape)
```

**Check the original text**

```python
# Display the first original movie review
print(df["text"].iloc[0])

# Display the sentiment label of the first review
print(df["label"].iloc[0])

# Display the column names
print(df.columns)
```

**Convert text to lowercase**

```python
# Store an example sentence in a variable
text = "THIS Movie Is AMAZING!"

# Convert all characters to lowercase
text = text.lower()

# Display the lowercase text
print(text)
```

**Remove HTML tags**

```python
# Store a sentence containing HTML tags
text = "This movie is <br> really good"

# Remove HTML tags using a regular expression
text = re.sub(r"<.*?>", " ", text)

# Display the cleaned sentence
print(text)
```

**Remove URLs**

```python
# Store a sentence containing a URL
text = "This movie is great https://example.com"

# Remove URLs from the sentence
text = re.sub(r"http\S+|www\.\S+", " ", text)

# Display the sentence without URLs
print(text)
```

**Remove punctuation**

```python
# Store a sentence containing punctuation marks
text = "Wow! This movie, is amazing."

# Replace punctuation marks with spaces
text = text.translate(
    str.maketrans(string.punctuation, " " * len(string.punctuation))
)

# Display the sentence without punctuation
print(text)
```

**Remove numbers**

```python
# Store a sentence containing numbers
text = "This movie has 10 stars and 2 actors"

# Replace numeric digits with spaces
text = re.sub(r"\d+", " ", text)

# Display the sentence without numbers
print(text)
```

**Tokenization**

```python
# Import the word tokenizer from NLTK
from nltk.tokenize import word_tokenize

# Download the tokenizer resources
nltk.download("punkt")
nltk.download("punkt_tab")

# Store a sentence for tokenization
text = "This movie is really amazing"

# Split the sentence into individual words
tokens = word_tokenize(text)

# Display the list of tokens
print(tokens)
```

**Remove stopwords**

```python
# Import the English stopwords list
from nltk.corpus import stopwords

# Download the stopwords resource
nltk.download("stopwords")

# Load the English stopwords into a set
stop_words = set(stopwords.words("english"))

# Preserve important negation words for sentiment analysis
stop_words = stop_words - {"not", "no", "nor"}

# Store example tokens
tokens = ["this", "movie", "is", "not", "really", "good"]

# Remove stopwords while keeping important negation words
filtered_tokens = [
    word for word in tokens if word not in stop_words
]

# Display the remaining tokens
print(filtered_tokens)
```

**Stemming**

```python
# Import the Porter Stemmer
from nltk.stem import PorterStemmer

# Create a stemming object
stemmer = PorterStemmer()

# Store words with different forms
words = ["playing", "played", "plays", "studies"]

# Convert each word into its stem
stemmed_words = [stemmer.stem(word) for word in words]

# Display the stemmed words
print(stemmed_words)
```

**: Lemmatization**

```python
# Import the WordNet Lemmatizer
from nltk.stem import WordNetLemmatizer

# Download the WordNet dictionary
nltk.download("wordnet")

# Create a lemmatizer object
lemmatizer = WordNetLemmatizer()

# Store example words
words = ["cars", "playing", "studies"]

# Convert words into their dictionary base forms
lemmatized_words = [
    lemmatizer.lemmatize(word) for word in words
]

# Display the lemmatized words
print(lemmatized_words)
```

**One-Hot Encoding**

```python
# Import MultiLabelBinarizer for binary word representation
from sklearn.preprocessing import MultiLabelBinarizer

# Define example documents as lists of words
documents = [
    ["good", "movie"],
    ["bad", "movie"],
    ["good"]
]

# Create the One-Hot Encoder
encoder = MultiLabelBinarizer()

# Convert word presence into binary values
encoded_data = encoder.fit_transform(documents)

# Convert the encoded matrix into a readable DataFrame
encoded_df = pd.DataFrame(
    encoded_data,
    columns=encoder.classes_
)

# Display the One-Hot Encoding result
print(encoded_df)
```

**Bag of Words (BoW)**

```python
# Import CountVectorizer for Bag of Words
from sklearn.feature_extraction.text import CountVectorizer

# Define example documents
documents = [
    "good movie good",
    "bad movie",
    "good film"
]

# Create the Bag of Words vectorizer
vectorizer = CountVectorizer()

# Learn the vocabulary and count word frequencies
bow_matrix = vectorizer.fit_transform(documents)

# Convert the sparse matrix into a readable DataFrame
bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

# Display the Bag of Words matrix
print(bow_df)
```
